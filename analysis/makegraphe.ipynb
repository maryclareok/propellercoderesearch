```python
import matplotlib.pyplot as plt
samples = []
timing_us = []
```

```python
f= open('single_cog_timing_values.txt','r')
for row in f:
    row = row.split(' ')
    samples.append(row[1])
    timing_us.append(row[3])
print(samples)
print(timing_us)
```

```python
samples = []
timing_us = []

with open("single_cog_timing_values.txt", "r") as f:
    for row in f:
        parts = row.strip().split()

        sample_number = int(parts[1].split("=")[1])
        time_value = int(parts[3].split("=")[1])

        samples.append(sample_number)
        timing_us.append(time_value)

print(samples)
print(timing_us)
```

```python
plt.title("single cog timing")
plt.xlabel("samples")
plt.ylabel("timing")
plt.plot(samples,timing_us,marker = 'o',c = 'g')
plt.show()
```

```python
samples2 = []
timing_us2 = []

with open("two_cog_timing_values.txt", "r") as f:
    for row2 in f:
        parts2 = row2.strip().split()

        sample_number2 = int(parts2[1].split("=")[1])
        time_value2 = int(parts2[3].split("=")[1])

        samples2.append(sample_number2)
        timing_us2.append(time_value2)

print(samples2)
print(timing_us2)
```

```python
plt.title("two cog timing")
plt.xlabel("samples")
plt.ylabel("timing")
plt.plot(samples2,timing_us2,marker = 'o',c = 'g')
plt.show()
```

```python
plt.title("single and two cog timing")
plt.xlabel("samples")
plt.ylabel("timing")
plt.plot(samples[:100], timing_us[:100],label = "single cog ")
plt.plot(samples2[:100], timing_us2[:100],label = "two cog")
plt.legend(loc="upper right")
plt.show()
```
